## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def suscriptores(n, memo=None):
    """
    Calcula el número de nuevos suscriptores en el mes n usando memoización.

    Parámetros:
        n (int): mes a consultar (n >= 0)
        memo (dict): diccionario de memoización (uso interno/recursivo)

    Retorna:
        int: número de nuevos suscriptores en el mes n
    """
    if memo is None:
        memo = {}

    # Paso 1: caso base
    if n <= 1:
        return n

    # Paso 2: ¿ya está calculado?
    if n in memo:
        return memo[n]

    # Paso 3: calcular, guardar y retornar
    resultado = suscriptores(n - 1, memo) + suscriptores(n - 2, memo)
    memo[n] = resultado
    return resultado


def secuencia_suscriptores(n):
    """Retorna la lista de T(0) a T(n), útil para ver la tendencia completa."""
    memo = {}
    return [suscriptores(i, memo) for i in range(n + 1)]


if __name__ == "__main__":
    # Ejemplo de uso
    mes = int(input("Ingrese el mes n a consultar: "))

    resultado = suscriptores(mes)
    print(f"\nNuevos suscriptores en el mes {mes}: {resultado}")

    print(f"\nSecuencia completa desde el mes 0 hasta el mes {mes}:")
    print(secuencia_suscriptores(mes))
```
